# VTX Data Exploration

Initial exploration of IMU data from VTX files.

**File**: Cornering Clockwise 1.vtx - 4 right-hand corners around a block

**Goals**:
1. Load and inspect the data
2. Visualize raw sensor readings
3. Identify which axes correspond to forward/lateral/vertical
4. Check data quality

**Note**: This is test data from uncalibrated hardware, so we're mainly validating the analysis pipeline.

```python
import sys
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Add scripts to path
sys.path.insert(0, os.path.join(os.getcwd(), '../scripts'))

from load_vtx import load_vtx_file

# Configure plotting
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")
%matplotlib inline
```

## 1. Load Data

```python
# Load the VTX file
data = load_vtx_file('../data/sample-recordings/cornering_clockwise_1.vtx')

header = data['header']
metadata = data['metadata']
df = data['samples']

print("=== FILE INFO ===")
print(f"Version: {header['version'][0]}.{header['version'][1]}")
print(f"Sample count: {header['sample_count']:,}")
print(f"Header sample rate: {header['sample_rate']:.1f} Hz")
print(f"Computed sample rate: {header['computed_sample_rate']:.1f} Hz")
print(f"Duration: {header['duration']:.2f} seconds")
print(f"\nColumns: {list(df.columns)}")
print(f"\n=== METADATA ===")
print(f"Device: {metadata.get('device', {}).get('name', 'Unknown')}")
print(f"Session: {metadata.get('session', {}).get('createdAt', 'Unknown')}")
```

```python
# Show first few rows
df.head(10)
```

## 2. Basic Statistics

Calculate mean and standard deviation for each axis to identify patterns.

```python
# Accelerometer statistics
print("=== ACCELEROMETER (m/s²) ===")
accel_cols = ['accel_x', 'accel_y', 'accel_z']
accel_stats = df[accel_cols].describe()
print(accel_stats)

print("\n=== GYROSCOPE (rad/s) ===")
gyro_cols = ['gyro_x', 'gyro_y', 'gyro_z']
gyro_stats = df[gyro_cols].describe()
print(gyro_stats)

if 'mag_x' in df.columns:
    print("\n=== MAGNETOMETER (µT) ===")
    mag_cols = ['mag_x', 'mag_y', 'mag_z']
    mag_stats = df[mag_cols].describe()
    print(mag_stats)
```

## 3. Identify Axes

Use statistical analysis to identify which axis corresponds to:
- **Vertical**: Should have mean close to ±9.8 m/s² (gravity)
- **Forward**: Should have highest variance (most motion during cycling)
- **Lateral**: Remaining axis

```python
# Calculate metrics for axis identification
gravity = 9.8

axes_analysis = []
for axis in ['x', 'y', 'z']:
    col = f'accel_{axis}'
    mean_val = df[col].mean()
    variance = df[col].var()
    gravity_distance = abs(abs(mean_val) - gravity)
    
    axes_analysis.append({
        'axis': axis,
        'mean': mean_val,
        'variance': variance,
        'std': df[col].std(),
        'gravity_distance': gravity_distance
    })

axes_df = pd.DataFrame(axes_analysis)
print("=== AXIS ANALYSIS ===")
print(axes_df)

# Identify axes
vertical_axis = axes_df.loc[axes_df['gravity_distance'].idxmin(), 'axis']
remaining = axes_df[axes_df['axis'] != vertical_axis]
forward_axis = remaining.loc[remaining['variance'].idxmax(), 'axis']
lateral_axis = remaining[remaining['axis'] != forward_axis].iloc[0]['axis']

print(f"\n=== IDENTIFIED AXES ===")
print(f"Vertical (gravity): {vertical_axis.upper()} (mean: {axes_df[axes_df['axis']==vertical_axis]['mean'].iloc[0]:.3f} m/s²)")
print(f"Forward (most motion): {forward_axis.upper()} (variance: {axes_df[axes_df['axis']==forward_axis]['variance'].iloc[0]:.3f})")
print(f"Lateral (remaining): {lateral_axis.upper()}")
```

## 4. Visualize Raw Data

Plot all axes over time to see patterns.

```python
fig, axes = plt.subplots(3, 1, figsize=(14, 10), sharex=True)

# Accelerometer
for axis in ['x', 'y', 'z']:
    axes[0].plot(df['time_sec'], df[f'accel_{axis}'], label=f'{axis.upper()} axis', alpha=0.7)
axes[0].axhline(y=9.8, color='r', linestyle='--', alpha=0.3, label='Gravity (9.8 m/s²)')
axes[0].axhline(y=-9.8, color='r', linestyle='--', alpha=0.3)
axes[0].set_ylabel('Acceleration (m/s²)')
axes[0].set_title('Accelerometer - All Axes')
axes[0].legend(loc='upper right')
axes[0].grid(True, alpha=0.3)

# Gyroscope
for axis in ['x', 'y', 'z']:
    axes[1].plot(df['time_sec'], df[f'gyro_{axis}'], label=f'{axis.upper()} axis', alpha=0.7)
axes[1].set_ylabel('Angular Velocity (rad/s)')
axes[1].set_title('Gyroscope - All Axes')
axes[1].legend(loc='upper right')
axes[1].grid(True, alpha=0.3)

# Magnetometer (if available)
if 'mag_x' in df.columns:
    for axis in ['x', 'y', 'z']:
        axes[2].plot(df['time_sec'], df[f'mag_{axis}'], label=f'{axis.upper()} axis', alpha=0.7)
    axes[2].set_ylabel('Magnetic Field (µT)')
    axes[2].set_title('Magnetometer - All Axes')
    axes[2].legend(loc='upper right')
    axes[2].grid(True, alpha=0.3)
else:
    axes[2].text(0.5, 0.5, 'No magnetometer data', ha='center', va='center')
    axes[2].set_title('Magnetometer - Not Available')

axes[2].set_xlabel('Time (seconds)')
plt.tight_layout()
plt.show()
```

## 5. Look for Corners

This file should have 4 right-hand corners. Let's see if we can spot them in the lateral acceleration.

```python
# Plot lateral acceleration (the axis that should show cornering)
lateral_col = f'accel_{lateral_axis}'

plt.figure(figsize=(14, 5))
plt.plot(df['time_sec'], df[lateral_col], label=f'Lateral Accel ({lateral_axis.upper()})', linewidth=0.8)
plt.axhline(y=0, color='k', linestyle='-', alpha=0.2)
plt.xlabel('Time (seconds)')
plt.ylabel('Lateral Acceleration (m/s²)')
plt.title(f'Lateral Acceleration - Looking for 4 Corners')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

print(f"Lateral acceleration stats:")
print(f"  Mean: {df[lateral_col].mean():.3f} m/s²")
print(f"  Std: {df[lateral_col].std():.3f} m/s²")
print(f"  Min: {df[lateral_col].min():.3f} m/s²")
print(f"  Max: {df[lateral_col].max():.3f} m/s²")
```

## 6. Data Quality Check

Check for:
- Missing or NaN values
- Sensor saturation (values at limits)
- Gaps in timestamps

```python
print("=== DATA QUALITY ===")

# Check for NaN
nan_counts = df.isna().sum()
if nan_counts.sum() > 0:
    print("\nNaN values found:")
    print(nan_counts[nan_counts > 0])
else:
    print("✓ No NaN values")

# Check timestamp gaps
time_diffs = df['time_sec'].diff().dropna()
expected_diff = 1.0 / header['computed_sample_rate']
large_gaps = time_diffs[time_diffs > expected_diff * 2]

if len(large_gaps) > 0:
    print(f"\n⚠ Found {len(large_gaps)} large timestamp gaps (> {expected_diff*2:.3f}s)")
    print(f"Largest gap: {large_gaps.max():.3f}s at index {large_gaps.idxmax()}")
else:
    print(f"✓ No large timestamp gaps (expected ~{expected_diff*1000:.1f}ms between samples)")

# Check for sensor saturation (typical IMU ranges)
print("\n=== SENSOR RANGE CHECK ===")
print(f"Accel range: ±{df[accel_cols].abs().max().max():.1f} m/s² (typical limit: ±156 m/s² for ±16g)")
print(f"Gyro range: ±{df[gyro_cols].abs().max().max():.1f} rad/s (typical limit: ±34.9 rad/s for ±2000°/s)")
if 'mag_x' in df.columns:
    print(f"Mag range: ±{df[mag_cols].abs().max().max():.1f} µT (typical: 25-65 µT)")
```
